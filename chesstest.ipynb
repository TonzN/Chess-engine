```python
import chess

def get_piece_from_uci_position(board, uci_position):
    # Convert the UCI position to a move object
    move = chess.Move.from_uci(uci_position)

    # Get the piece on the starting square of the move
    piece = board.piece_at(move.from_square)

    return piece

# Example usage:
board = chess.Board()
uci_position = "e2e4"
piece = get_piece_from_uci_position(board, uci_position)

if piece:
    print(f"The piece on {uci_position} is {piece.symbol()} and its team is {'White' if piece.color == chess.WHITE else 'Black'}")
else:
    print(f"No piece on {uci_position}")
```
